## 2. Class Distribution & Missing Values

```python
# Count records per class (0 = HAM / Legitimate, 1 = SPAM)
df["label"].value_counts()
```

```python
# Proportion / percentage of HAM vs SPAM
df["label"].value_counts(normalize=True)
```

```python
# Check for missing/null values across columns
df.isnull().sum()
```

## 3. Duplicate Analysis & Investigation

```python
# Count duplicate email text contents
df["email"].duplicated().sum()
```

```python
# Check for duplicate filenames
df["filename"].duplicated().sum()
```

above showed that dataset contains the duplicate emails . we need to invistigate this

```python
# Total duplicate rows across all columns
df.duplicated().sum()
```

```python
# Filter and inspect rows that share identical email text
duplicate_emails = df[df["email"].duplicated(keep=False)]

duplicate_emails.head(10)
```

```python
# View label, email, and filename for duplicate records
duplicate_emails[["label", "email", "filename"]].head(20)
```

```python
# Verify whether identical emails ever have conflicting labels (both ham and spam)
duplicate_label_check = (
    df.groupby("email")["label"]
      .nunique()
)

duplicate_label_check.value_counts()
```

```python
# Re-check class distribution counts
df["label"].value_counts()
```

## 4. Email Length & Repetition Statistics

```python
# Compute initial character count and word count per email and view statistical summary
df["char_count"] = df["email"].str.len()
df["word_count"] = df["email"].str.split().str.len()

df[["char_count", "word_count"]].describe()
```

```python
# Number of unique email texts
df["email"].nunique()
```

```python
# How many rows are actually repeated?
len(df) - df["email"].nunique()
```

```python
# Frequency distribution: how many times duplicate emails are repeated
duplicate_frequency = df["email"].value_counts()

duplicate_frequency.value_counts().sort_index()
```
