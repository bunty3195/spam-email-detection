# 05 - Model Training
In this notebook, we train machine learning classification models using the TF-IDF feature matrix.

### Planned Workflow:
1. **Prepare Features & Target**:
   - `X`: TF-IDF feature matrix from notebook 04
   - `y`: Target labels (0 = HAM, 1 = SPAM)
2. **Train / Test Split**:
   - Stratified train-test split (e.g., 80% train, 20% test) to preserve class balance
3. **Train Classification Models**:
   - **Multinomial Naive Bayes** (`MultinomialNB`) - Classic benchmark for text classification
   - **Logistic Regression** (`LogisticRegression`) - Strong linear baseline
   - **Support Vector Classifier** (`LinearSVC`) - High accuracy in high-dimensional sparse spaces
4. **Save Models**:
   - Save trained models and vectorizer to `models/` directory for inference
