# 02 - Data Cleaning
Data cleaning pipeline based on the findings from dataset analysis:
1. **Remove Duplicate Emails**: Deduplicate emails to avoid data leakage and bias
2. **Normalize Whitespace**: Strip extraneous spaces, newlines, and tabs using regex
3. **Recalculate Length Metrics**: Recompute character and word counts on clean text
4. **Save Cleaned Dataset**: Export cleaned data for downstream EDA and Modeling

## 1. Load Raw Dataset

```python
import pandas as pd

# Load the raw dataset to begin cleaning
# Note: If running from notebooks directory, path may be "../data/raw/ENRON-spam.csv"
df = pd.read_csv("Enron_Txt_fn_orignal.csv")
print("Initial records:", len(df))
```

## 2. Remove Duplicate Emails

```python
# Create a copy with duplicate email texts removed (keeping the first occurrence)
df_clean = df.drop_duplicates(subset="email").copy()
```

```python
print("Original:", len(df))
print("After removing duplicate emails:", len(df_clean))
print("Removed:", len(df) - len(df_clean))
```

```python
# Deduplicate the main dataframe and verify removed count
original_count = len(df)

df = df.drop_duplicates(subset="email").copy()

cleaned_count = len(df)

print("Original records:", original_count)
print("Unique records:", cleaned_count)
print("Duplicates removed:", original_count - cleaned_count)
```

```python
#stored the "df"veriable into new clean dataset ( remove the duplicates )
df = df.drop_duplicates(subset="email").copy()
```

```python
# Check shape of dataframe after deduplication
df.shape
```

```python
# Inspect first 10 deduplicated records
df.head(10)
```

## 3. Normalize Whitespace
Clean repetitive whitespaces, tabs, and newlines from raw email text.

```python
"""Normalize whitespace """

""" This changes:

Subject: hello

How are you?

Thanks

into:

Subject: hello How are you? Thanks          """
```

```python
# Replace multiple consecutive whitespace characters with a single space and strip borders
import re

df["email"] = df["email"].apply(
    lambda x: re.sub(r"\s+", " ", x).strip()
)
```

```python
# Recalculate character count on cleaned text
df["char_count"] = df["email"].str.len()
```

```python
# Recalculate word count on cleaned text
df["word_count"] = df["email"].str.split().str.len()
```

```python
# Statistical summary of character and word counts after cleaning
df[["char_count", "word_count"]].describe()
```
